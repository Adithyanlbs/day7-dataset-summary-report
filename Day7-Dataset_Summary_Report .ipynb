# Day 7 – Dataset Summary Report

## Pandas Dataset Exploration

This notebook uses Python and Pandas to load and explore the provided student CSV dataset. It demonstrates the basic structure and characteristics of the dataset, including rows, columns, index, data types, descriptive statistics, and missing values.


## 1. Import Pandas

```python
import pandas as pd
```

## 2. Load the CSV Dataset

The dataset is loaded into a Pandas DataFrame using `read_csv()`.


```python
df = pd.read_csv("Day7_Student_Dataset.csv")

print("Dataset loaded successfully.")
```

## 3. Inspect the First Few Rows – `head()`

`head()` displays the first five rows of the DataFrame by default.


```python
df.head()
```

## 4. Inspect the Last Few Rows – `tail()`

`tail()` displays the last five rows of the DataFrame by default.


```python
df.tail()
```

## 5. View Random Samples – `sample()`

`sample()` returns randomly selected rows from the dataset.


```python
df.sample(5, random_state=42)
```

## 6. Check Dataset Shape

`shape` returns the number of rows and columns in the form `(rows, columns)`.


```python
print("Shape:", df.shape)
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

## 7. View Column Names – `columns`

`columns` displays the names of all features/columns in the DataFrame.


```python
print("Columns:")
print(df.columns.tolist())
```

## 8. View the DataFrame Index – `index`

`index` shows the row labels/index of the DataFrame.


```python
print("Index:")
print(df.index)
```

## 9. Check Data Types – `dtypes`

`dtypes` shows the data type of each column.


```python
print(df.dtypes)
```

## 10. Dataset Information – `info()`

`info()` provides a concise summary including the number of entries, column names, non-null counts, and data types.


```python
df.info()
```

## 11. Descriptive Statistics – `describe()`

`describe()` provides basic statistical information for numeric columns, including count, mean, standard deviation, minimum, quartiles, and maximum.


```python
df.describe()
```

## 12. Missing Value Analysis

`isna()` identifies missing values. `sum()` counts the missing values in each column.


```python
missing_values = df.isna().sum()

print("Missing values in each column:")
print(missing_values)
```

## 13. Check Whether the Dataset Contains Any Missing Values

```python
total_missing = df.isna().sum().sum()

print("Total missing values:", total_missing)

if total_missing == 0:
    print("No missing values were found in the dataset.")
else:
    print("The dataset contains missing values.")
```

## 14. Data Types Summary

```python
print("Numeric columns:")
print(df.select_dtypes(include="number").columns.tolist())

print("\nNon-numeric columns:")
print(df.select_dtypes(exclude="number").columns.tolist())
```
